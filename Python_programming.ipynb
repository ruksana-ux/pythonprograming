**pyramid pattern**

```python
rows = 5
# to generate upper part of the diamond
for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ", end=" ")
    for k in range(1, 2 * i):
        print("*", end=" ")
    print()

# Lower part of diamond
for i in range(rows - 1, 0, -1):
    for j in range(rows - i):
        print(" ", end=" ")
    for k in range(1, 2 * i):
        print("*", end=" ")
    print()
```

**to convert temperatures from Fahrenheit to Celsius several times in programme**

```python
temperature1 = 77
celsius1 = (temperature1 - 32) * 5 / 9
print(celsius1)

temperature2 = 95
celsius2 = (temperature2 - 32) * 5 / 9
print(celsius2)

temperature3 = 50
celsius3 = (temperature3 - 32) * 5 / 9
print(celsius3)
```

```python
def fahrenheit_to_celsius(fahrenheit):
  return (fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celsius(77))
print(fahrenheit_to_celsius(95))
print(fahrenheit_to_celsius(50))
```

**Python Function Arguments**

* An argument is the actual value that is sent to the function when it is called.

```python
# name is a parameter
def my_function(name):
  print("Hello", name)

my_function("Ruksana")
```

**to default value**

```python
def my_function(country = "India"):
  print("I am from", country)

my_function("Sweden")
my_function("Canada")
my_function()
my_function("Brazil")
```

**Keyword Arguments**

```python
def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

my_function(animal = "dog", name = "Green")
```

```python
def my_function(sports, name):
  print("My favourite sport is", sports)
  print("My", sports + "'s favourite player name is", name)

my_function(name = "Saina Nehwal", sports = "batmiton")
```

**Positional Arguments**

```python
def my_function(sports, name):
  print("My favourite sport is ", sports)
  print("My", sports + "'s favourite player name is", name)

my_function("Batminton", "Saina Nehwal")
```

*args and **kwargs

```python
def my_function(*brothers):
  print("My youngest brother is " + brothers[2])

my_function("Arman", "Abel", "Aman")
```

```python
def my_function(*numbers):
  total = 0
  for num in numbers:
    total += num
  return total

print(my_function(1, 2, 3))
print(my_function(10, 20, 30, 40))
print(my_function(5))
```

**kwargs

* The **kwargs parameter allows a function to accept any number of keyword arguments.

```python
def my_function(**myvar):
  print("Type:", type(myvar))
  print("Name:", myvar["name"])
  print("Age:", myvar["age"])
  print("All data:", myvar)

my_function(name = "Reyan", age = 35, city = "Delhi")
```
