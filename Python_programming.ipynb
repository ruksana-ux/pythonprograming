**Rightangle Triangle pattern programming**

```python
for i in range(1,6):
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```

**Inverted L shape**

```python
for i in range(1, 6):
    for j in range(1, 6):
        if i == 1 or i == 6 or j == 1 or j == 6 :
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

**Square shaped pattern using else statement**

```python
for i in range(1, 6):
    for j in range(1, 6):
        if i == 1 or i == 6-1 or j ==1 or j == 6-1:
            print("*", end =" ")
        else:
            print(" ", end =" ")
    print()
```

**pyramid pattern**

```python
rows = 5
# to generate upper part of the diamond
for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ", end=" ")
    for k in range(1, 2 * i):
        print("*", end=" ")
    print()

# Lower part of diamond
for i in range(rows - 1, 0, -1):
    for j in range(rows - i):
        print(" ", end=" ")
    for k in range(1, 2 * i):
        print("*", end=" ")
    print()
```
